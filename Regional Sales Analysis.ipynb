## 🧹 Data Cleaning & Wrangling

```python
# 🔧 4. XỬ LÝ VÀ KẾT HỢP DỮ LIỆU

# 4.1 Kết hợp dữ liệu với bảng Customers
df = df_sales.merge(
    df_customers,
    how='left',
    left_on='Customer Name Index',
    right_on='Customer Index'
)

# 4.2 Kết hợp dữ liệu với bảng Products
df = df.merge(
    df_products,
    how='left',
    left_on='Product Description Index',
    right_on='Index'
)

# 4.3 Kết hợp dữ liệu với bảng Regions
df = df.merge(
    df_regions,
    how='left',
    left_on='Delivery Region Index',
    right_on='id'
)

# 4.4 Kết hợp dữ liệu với bảng State Regions
df = df.merge(
    df_state_reg[["State Code", "Region"]],
    how='left',
    left_on='state_code',
    right_on='State Code'
)

# 4.5 Kết hợp dữ liệu với bảng Budgets
df = df.merge(
    df_budgets,
    how='left',
    on='Product Name'
)

# 4.6 Xóa các cột dư thừa sau khi kết hợp
cols_to_drop = [
    'Customer Index',
    'Index',
    'id',
    'State Code'
]

df = df.drop(columns=cols_to_drop, errors='ignore')

# Xem 5 dòng đầu tiên của DataFrame sau khi kết hợp
df.head()
```

```python
# Chuyển tất cả tên cột sang chữ thường để thống nhất và dễ truy cập
df.columns = df.columns.str.lower()

# Hiển thị danh sách tên cột sau khi cập nhật
df.columns.values
```

```python
# 🔧 4.x — XÓA CÁC CỘT KHÔNG CẦN THIẾT VÀ ĐỔI TÊN CỘT CHO DỄ HIỂU
# ------------------------------------------------------------------

# 1) Chỉ giữ lại các cột cần thiết, bao gồm cả 'region' (khu vực)
cols_to_keep = [
    'ordernumber',        # Mã đơn hàng duy nhất
    'orderdate',          # Ngày đặt hàng
    'customer names',     # Tên khách hàng đặt hàng
    'channel',            # Kênh bán hàng (ví dụ: Wholesale, Distributor)
    'product name',       # Tên sản phẩm được mua
    'order quantity',     # Số lượng sản phẩm được đặt
    'unit price',          # Giá bán trên mỗi đơn vị
    'line total',          # Doanh thu của dòng sản phẩm (số lượng × đơn giá)
    'total unit cost',     # Tổng chi phí của dòng sản phẩm
    'state_code',          # Mã viết tắt gồm 2 ký tự của bang
    'state',               # Tên đầy đủ của bang
    'region',              # Khu vực rộng hơn của Hoa Kỳ (ví dụ: South, West)
    'latitude',             # Vĩ độ của thành phố giao hàng
    'longitude',            # Kinh độ của thành phố giao hàng
    '2017 budgets'         # Mục tiêu ngân sách của sản phẩm trong năm 2017
]

# Chỉ giữ lại các cột cần thiết
df = df[cols_to_keep]


# 2) Đổi tên các cột thành tên ngắn gọn, rõ ràng và nhất quán
df = df.rename(columns={
    'ordernumber':     'order_number',    # Mã đơn hàng
    'orderdate':       'order_date',      # Ngày đặt hàng
    'customer names':  'customer_name',   # Tên khách hàng
    'product name':    'product_name',    # Tên sản phẩm
    'order quantity':  'quantity',        # Số lượng sản phẩm bán ra
    'unit price':       'unit_price',     # Đơn giá bán (USD)
    'line total':       'revenue',        # Doanh thu của dòng sản phẩm
    'total unit cost':  'cost',           # Chi phí của dòng sản phẩm
    'state_code':       'state',          # Mã viết tắt của bang
    'state':            'state_name',     # Tên đầy đủ của bang
    'region':           'us_region',      # Khu vực của Hoa Kỳ
    'latitude':         'lat',            # Vĩ độ
    'longitude':        'lon',            # Kinh độ
    '2017 budgets':     'budget'          # Mục tiêu ngân sách năm 2017
})

# Hiển thị 5 dòng đầu tiên của DataFrame sau khi làm sạch
df.head()
```

## 🛠 Feature Engineering

```python
# 🧮 5. TẠO CÁC BIẾN PHÂN TÍCH

# 1. Tính tổng chi phí cho từng dòng sản phẩm
df['total_cost'] = df['quantity'] * df['cost']

# 2. Tính lợi nhuận = doanh thu - tổng chi phí
df['profit'] = df['revenue'] - df['total_cost']

# 3. Tính tỷ suất lợi nhuận (%)
df['profit_margin_pct'] = (df['profit'] / df['revenue']) * 100

# 4. Lấy tên đầy đủ của tháng từ order_date để hiển thị
# Ví dụ: January, February, March
df['order_month_name'] = df['order_date'].dt.month_name()

# 5. Lấy số thứ tự của tháng để sắp xếp đúng thứ tự từ 1–12
df['order_month_num'] = df['order_date'].dt.month

# 6. (Tùy chọn) Hiển thị toàn bộ DataFrame sau khi cập nhật
df
```

## 🔹 1. XU HƯỚNG DOANH THU THEO THÁNG

**Mục tiêu:** Theo dõi xu hướng doanh thu theo thời gian để phát hiện tính mùa vụ hoặc các giai đoạn doanh thu tăng đột biến.

**Biểu đồ:** Biểu đồ đường (Line Chart)

**Loại phân tích EDA:** Phân tích theo thời gian (Time Series)

**Cấu trúc biểu đồ:** Sử dụng đường biểu diễn kết hợp với các điểm đánh dấu (markers) để thể hiện rõ doanh thu của từng tháng.

```python
# Chuyển order_date sang dạng kỳ theo tháng
# Ví dụ: 2022-01, 2022-02
df['order_month'] = df['order_date'].dt.to_period('M')

# Tính tổng doanh thu theo từng tháng
monthly_sales = df.groupby('order_month')['revenue'].sum()

# Thiết lập kích thước biểu đồ để dễ quan sát
plt.figure(figsize=(15, 4))

# Vẽ xu hướng doanh thu theo tháng
# Sử dụng điểm đánh dấu hình tròn (marker) cho từng tháng
monthly_sales.plot(marker='o', color='navy')

# Chuyển đơn vị trục Y sang triệu (Millions) để dễ đọc
from matplotlib.ticker import FuncFormatter

formatter = FuncFormatter(lambda x, pos: f'{x/1e6:.1f}M')
plt.gca().yaxis.set_major_formatter(formatter)

# Thêm tiêu đề và nhãn cho các trục
plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Total Revenue (Millions)')

# Xoay nhãn trên trục X 45 độ để dễ đọc
plt.xticks(rotation=45)

# Tự động điều chỉnh bố cục để tránh nội dung bị cắt
plt.tight_layout()

# Hiển thị biểu đồ
plt.show()
```

### 💡 INSIGHTS

- Doanh thu có xu hướng dao động ổn định trong khoảng **24M–26M**, với các đỉnh doanh thu rõ rệt vào cuối mùa xuân đến đầu mùa hè (**tháng 5–6**) và mức thấp nhất thường xuất hiện vào **tháng 1**.

- Xu hướng doanh thu nhìn chung **ổn định qua các năm**, cho thấy nhu cầu thị trường có tính **mùa vụ tương đối rõ ràng và nhất quán**.

- Tuy nhiên, mức **sụt giảm doanh thu mạnh vào đầu năm 2017** là một điểm bất thường đáng chú ý. Cần phân tích sâu hơn để xác định nguyên nhân, chẳng hạn như **biến động thị trường, vấn đề trong hoạt động kinh doanh hoặc chương trình khuyến mãi không đúng thời điểm**.

## 🔹 2. XU HƯỚNG DOANH THU THEO THÁNG (TỔNG HỢP TẤT CẢ CÁC NĂM)

**🎯 Mục tiêu:** Phân tích xu hướng mùa vụ tổng thể bằng cách tổng hợp doanh thu của cùng một tháng trong tất cả các năm.

**📊 Biểu đồ:** Biểu đồ đường (Line Chart)

**📈 Loại phân tích EDA:** Phân tích theo thời gian (Time Series)

**📋 Cấu trúc biểu đồ:**
- Trục X: Các tháng từ **tháng 1 đến tháng 12**.
- Trục Y: Tổng doanh thu.
- Sử dụng **đường biểu diễn kết hợp với markers** để thể hiện rõ doanh thu của từng tháng.
- Các tháng được sắp xếp theo **số thứ tự từ 1 đến 12** nhằm phản ánh đúng chu kỳ thời gian.

```python
# 1. Loại bỏ các đơn hàng của năm 2018
df_ = df[df['order_date'].dt.year != 2018]

# Loại bỏ dữ liệu của năm chưa đầy đủ (ví dụ: tháng 1 và tháng 2 năm 2018)
# để tổng doanh thu theo tháng không bị sai lệch do dữ liệu không đầy đủ.
# Tạo df_ mới để không thay đổi DataFrame gốc,
# giữ nguyên df cho các phân tích khác và chỉ lọc dữ liệu cho biểu đồ này.


# 2. Nhóm theo số tháng và tên tháng, tính tổng doanh thu,
# sau đó sắp xếp theo thứ tự số tháng
monthly_sales = (
    df_
    .groupby(['order_month_num', 'order_month_name'])['revenue']
    .sum()
    .sort_index()
)


# 3. Thiết lập biểu đồ
from matplotlib.ticker import FuncFormatter

plt.figure(figsize=(13, 4))

plt.plot(
    monthly_sales.index.get_level_values(1),  # Trục X: tên tháng
    monthly_sales.values,                     # Trục Y: tổng doanh thu
    marker='o',                               # Điểm đánh dấu hình tròn
    color='navy'                              # Màu của đường biểu diễn
)


# 4. Chuyển giá trị trục Y sang đơn vị triệu để dễ đọc
formatter = FuncFormatter(lambda x, pos: f'{x/1e6:.1f}M')
plt.gca().yaxis.set_major_formatter(formatter)


# 5. Thêm tiêu đề và nhãn cho các trục
plt.title('Overall Monthly Sales Trend (Excluding 2018)')
plt.xlabel('Month')
plt.ylabel('Total Revenue (Millions)')


# 6. Xoay nhãn trên trục X để dễ đọc
plt.xticks(rotation=45)


# 7. Điều chỉnh bố cục để tránh nội dung bị cắt
plt.tight_layout()


# 8. Hiển thị biểu đồ
plt.show()
```

### 💡 INSIGHTS

- Qua tất cả các năm, **tháng 1** bắt đầu với doanh thu khá cao, khoảng **99M**, sau đó giảm mạnh và chạm mức thấp nhất vào **tháng 4** (≈ **95M**).
- Doanh thu phục hồi vào **tháng 5** và **tháng 8** (≈ **102M**), trước khi duy trì ở mức tương đối ổn định **99–101M** từ tháng 9 đến tháng 12.
- Mô hình này cho thấy **doanh thu tăng mạnh sau Tết**, giảm vào mùa xuân và có một **đợt tăng vào giữa mùa hè** trong mỗi năm.

```python
# Tính tổng doanh thu cho từng sản phẩm và chuyển giá trị sang đơn vị triệu
top_prod = df.groupby('product_name')['revenue'].sum() / 1_000_000

# Chọn 10 sản phẩm có doanh thu cao nhất
top_prod = top_prod.nlargest(10)

# Thiết lập kích thước biểu đồ để dễ quan sát
plt.figure(figsize=(9, 4))

# Vẽ biểu đồ thanh ngang: trục X là doanh thu tính bằng triệu, trục Y là tên sản phẩm
sns.barplot(
    x=top_prod.values,    # Trục X: giá trị doanh thu tính bằng triệu
    y=top_prod.index,     # Trục Y: tên sản phẩm
    palette='viridis'     # Bảng màu cho các thanh
)

# Thêm tiêu đề và nhãn cho các trục
plt.title('Top 10 Products by Revenue (in Millions)')  # Tiêu đề chính
plt.xlabel('Total Revenue (in Millions)')              # Nhãn trục X
plt.ylabel('Product Name')                             # Nhãn trục Y

# Điều chỉnh bố cục để tránh các thành phần bị chồng lấn
plt.tight_layout()

# Hiển thị biểu đồ
plt.show()
```

### 💡 INSIGHTS

- **Sản phẩm 26 và 25** dẫn đầu rõ rệt với doanh thu lần lượt khoảng **118M và 110M**, sau đó giảm mạnh xuống **78M** đối với **Sản phẩm 13**, trong khi nhóm sản phẩm ở mức trung bình tập trung trong khoảng **68–75M**.
- **4 sản phẩm có doanh thu thấp nhất** tập trung trong khoảng **52–57M**, cho thấy hiệu quả kinh doanh của nhóm này tương đối tương đồng.
- Nên tập trung vào **thử nghiệm các chiến lược tăng trưởng đối với nhóm sản phẩm trung bình** và **nâng cao hiệu quả hoạt động đối với nhóm có doanh thu thấp** để tạo ra mức tăng trưởng đáng kể.

## 🔹 4. TOP 10 SẢN PHẨM THEO TỶ SUẤT LỢI NHUẬN TRUNG BÌNH

**🎯 Mục tiêu:** So sánh **tỷ suất lợi nhuận trung bình** giữa các sản phẩm để xác định những sản phẩm có **biên lợi nhuận cao nhất**.

**📊 Biểu đồ:** Biểu đồ thanh ngang (Horizontal Bar Chart)

**📈 Loại phân tích EDA:** Phân tích đơn biến (Univariate Analysis)

**📋 Cấu trúc biểu đồ:**
- Các thanh được **sắp xếp theo thứ tự tỷ suất lợi nhuận trung bình giảm dần**.
- Chỉ hiển thị **Top 10 sản phẩm** có tỷ suất lợi nhuận trung bình cao nhất.
- Giá trị **Average Profit Margin** được sử dụng để so sánh khả năng sinh lời giữa các sản phẩm.

```python
# 1️⃣ Tính lợi nhuận trung bình cho từng sản phẩm và chọn Top 10
top_margin = (
    df.groupby('product_name')['profit']
      .mean()                        # Tính lợi nhuận trung bình cho từng sản phẩm
      .sort_values(ascending=False)  # Sắp xếp từ lợi nhuận trung bình cao nhất đến thấp nhất
      .head(10)                      # Chỉ giữ lại 10 sản phẩm đứng đầu
)

# 2️⃣ Thiết lập kích thước biểu đồ để dễ quan sát
plt.figure(figsize=(9, 4))

# 3️⃣ Vẽ biểu đồ thanh ngang
sns.barplot(
    x=top_margin.values,    # Trục X: giá trị lợi nhuận trung bình
    y=top_margin.index,     # Trục Y: tên sản phẩm
    palette='viridis'       # Bảng màu cho các thanh
)

# 4️⃣ Thêm tiêu đề và nhãn cho các trục
plt.title('Top 10 Products by Avg Profit Margin')  # Tiêu đề biểu đồ
plt.xlabel('Average Profit Margin (USD)')          # Nhãn trục X
plt.ylabel('Product Name')                         # Nhãn trục Y

# 5️⃣ Điều chỉnh bố cục để tránh nội dung bị cắt và hiển thị biểu đồ
plt.tight_layout()
plt.show()
```

- Các sản phẩm ở nhóm trung bình như **12, 26 và 21** tập trung trong khoảng **7.7–7.8K**, trong khi nhóm thấp nhất (**Sản phẩm 4, 16 và 1**) nằm trong khoảng **7.4–7.6K**.

## 🔹 5. DOANH THU THEO KÊNH (BIỂU ĐỒ TRÒN)

**🎯 Mục tiêu:** Thể hiện **cơ cấu tổng doanh thu theo từng kênh bán hàng** nhằm xác định các **kênh bán hàng chủ đạo**.

**📊 Biểu đồ:** Biểu đồ tròn (Pie Chart)

**📈 Loại phân tích EDA:** Phân tích đơn biến (Univariate Analysis)

**📋 Cấu trúc biểu đồ:**
- Các phần của biểu đồ thể hiện **tỷ trọng doanh thu của từng kênh bán hàng**.
- Hiển thị **phần trăm (%)** trên từng phần của biểu đồ.
- Sử dụng **màu sắc khác nhau** để dễ dàng phân biệt các kênh.
- Điều chỉnh **góc bắt đầu (Start Angle)** để bố cục biểu đồ dễ quan sát hơn.

```python
# Nhóm tổng doanh thu theo từng kênh bán hàng và sắp xếp theo thứ tự giảm dần
chan_sales = df.groupby('channel')['revenue'].sum().sort_values(ascending=False)

# Thiết lập kích thước biểu đồ tròn
plt.figure(figsize=(5, 5))

# Vẽ biểu đồ tròn với tỷ lệ phần trăm và góc bắt đầu được xác định
plt.pie(
    chan_sales.values,                   # Dữ liệu: giá trị doanh thu theo từng kênh
    labels=chan_sales.index,             # Nhãn: tên các kênh bán hàng
    autopct='%1.1f%%',                   # Hiển thị tỷ lệ phần trăm với 1 chữ số thập phân
    startangle=140,                      # Xoay biểu đồ để phần đầu tiên bắt đầu tại góc 140 độ
    colors=sns.color_palette('coolwarm') # Bảng màu cho các phần của biểu đồ
)

# Thêm tiêu đề để cung cấp thông tin ngữ cảnh
plt.title('Total Sales by Channel')

# Điều chỉnh bố cục để đảm bảo tất cả thành phần hiển thị vừa vặn
plt.tight_layout()

# Hiển thị biểu đồ
plt.show()
```

### 💡 INSIGHTS

- **Wholesale** chiếm khoảng **54% doanh thu**, tiếp theo là **Distributors** với khoảng **31%** và **Exports** với khoảng **15%**, cho thấy doanh nghiệp đang **phụ thuộc khá nhiều vào các kênh bán buôn nội địa**.

- Để **đa dạng hóa nguồn doanh thu và giảm rủi ro tập trung**, doanh nghiệp nên ưu tiên **mở rộng hoạt động xuất khẩu**, thông qua **marketing hướng đến thị trường quốc tế** và xây dựng **quan hệ hợp tác chiến lược với các đối tác nước ngoài**.

## 🔹 6. PHÂN PHỐI GIÁ TRỊ ĐƠN HÀNG TRUNG BÌNH (AOV)

**🎯 Mục tiêu:** Phân tích **phân phối giá trị đơn hàng** để xác định **mức chi tiêu phổ biến** và phát hiện các **giá trị đơn hàng bất thường (Outliers)**.

**📊 Biểu đồ:** Histogram

**📈 Loại phân tích EDA:** Phân tích đơn biến (Univariate Analysis)

**📋 Cấu trúc biểu đồ:**
- Sử dụng **50 bins** để phân chia các mức giá trị đơn hàng.
- Các cột biểu đồ được **tô màu và làm nổi bật viền** để dễ quan sát.
- Trục Y thể hiện **tần suất xuất hiện** của các giá trị đơn hàng.
- Biểu đồ giúp xác định **mức giá trị đơn hàng phổ biến**, độ phân tán và các **giá trị bất thường**.

```python
# Tính tổng doanh thu cho từng đơn hàng để xác định giá trị đơn hàng
aov = df.groupby('order_number')['revenue'].sum()

# Thiết lập kích thước biểu đồ để dễ quan sát hơn
plt.figure(figsize=(12, 4))

# Vẽ biểu đồ Histogram thể hiện giá trị các đơn hàng
plt.hist(
    aov,               # Dữ liệu: danh sách giá trị đơn hàng
    bins=50,           # Số lượng khoảng để phân nhóm giá trị đơn hàng
    color='skyblue',   # Màu nền của các cột
    edgecolor='black'  # Màu viền của các cột
)

# Thêm tiêu đề và nhãn cho các trục để cung cấp thông tin
plt.title('Distribution of Average Order Value')
plt.xlabel('Order Value (USD)')
plt.ylabel('Number of Orders')

# Điều chỉnh bố cục để tránh nội dung bị cắt
plt.tight_layout()

# Hiển thị biểu đồ
plt.show()
```

### 💡 INSIGHTS

- Phân phối **giá trị đơn hàng** có xu hướng **lệch phải rõ rệt (right-skewed)**, với phần lớn đơn hàng tập trung trong khoảng **20K–120K** và mức phổ biến nhất vào khoảng **50K–60K**.
- Một **đuôi dài (long tail)** của các giao dịch có giá trị cao kéo dài đến khoảng **400K–500K**. Tuy nhiên, các đơn hàng có giá trị lớn này chỉ chiếm **một tỷ trọng nhỏ trong tổng số đơn hàng**.

## 🔹 7. TỶ SUẤT LỢI NHUẬN (%) SO VỚI GIÁ BÁN ĐƠN VỊ

**🎯 Mục tiêu:** Phân tích **mối quan hệ giữa giá bán đơn vị (Unit Price)** và **tỷ suất lợi nhuận (%)** trên từng đơn hàng.

**📊 Biểu đồ:** Biểu đồ phân tán (Scatter Plot)

**📈 Loại phân tích EDA:** Phân tích hai biến (Bivariate Analysis)

**📋 Cấu trúc biểu đồ:**
- Mỗi điểm biểu diễn một **đơn hàng**.
- **Trục X:** Giá bán đơn vị (Unit Price).
- **Trục Y:** Tỷ suất lợi nhuận (%).
- Sử dụng **độ trong suốt (Transparency)** cho các điểm dữ liệu để thể hiện **mật độ dữ liệu** và hạn chế hiện tượng các điểm chồng lấn lên nhau.

```python
# Tính tỷ suất lợi nhuận (%) cho từng bản ghi
df['profit_margin_pct'] = (df['profit'] / df['revenue']) * 100

# Thiết lập kích thước biểu đồ để dễ quan sát
plt.figure(figsize=(6,4))

# Vẽ biểu đồ thể hiện mối quan hệ giữa giá bán đơn vị và tỷ suất lợi nhuận
plt.scatter(
    df['unit_price'],            # Trục X: giá bán đơn vị tính bằng USD
    df['profit_margin_pct'],     # Trục Y: tỷ suất lợi nhuận (%)
    alpha=0.6,                   # Mức độ trong suốt của các điểm khi bị chồng lấn
    color='green'                # Màu của các điểm dữ liệu
)

# Thêm tiêu đề và nhãn cho các trục
plt.title('Profit Margin % vs. Unit Price')  # Tiêu đề biểu đồ
plt.xlabel('Unit Price (USD)')                # Nhãn trục X
plt.ylabel('Profit Margin (%)')               # Nhãn trục Y

# Điều chỉnh bố cục để tránh nội dung bị cắt
plt.tight_layout()

# Hiển thị biểu đồ
plt.show()
```

### 💡 INSIGHTS

- Tỷ suất lợi nhuận tập trung chủ yếu trong khoảng **18%–60%**, không cho thấy mối tương quan rõ ràng với **giá bán đơn vị**, trong khi giá bán dao động từ gần **0 đến hơn 6,500 USD**.
- Các **dải điểm nằm ngang** với mật độ cao cho thấy các mức tỷ suất lợi nhuận tương đối ổn định trên một phạm vi giá rộng, phản ánh **chiến lược định giá tương đối nhất quán**.
- Các **giá trị ngoại lệ (Outliers)** dưới **18%** ở cả mức giá thấp và cao có thể cho thấy **sự kém hiệu quả về chi phí hoặc vấn đề trong định giá**, cần được phân tích sâu hơn.

## 🔹 8. PHÂN PHỐI GIÁ BÁN ĐƠN VỊ THEO SẢN PHẨM

## 🔹 9. TỔNG DOANH THU THEO KHU VỰC TẠI MỸ

**🎯 Mục tiêu:** So sánh **tổng doanh thu giữa các khu vực của Hoa Kỳ** để xác định **những thị trường có hiệu quả cao** và các khu vực cần tập trung để tăng trưởng.

**📊 Biểu đồ:** Biểu đồ thanh ngang (Horizontal Bar Chart)

**📈 Loại phân tích EDA:** Phân tích so sánh đơn biến (Univariate Comparison)

**📋 Cấu trúc biểu đồ:**
- Các thanh được **sắp xếp theo thứ tự tăng dần** (**Northeast → West**) để dễ dàng quan sát thứ hạng từ thấp đến cao.
- **Trục X:** Tổng doanh thu tính bằng **triệu USD**.
- **Trục Y:** Tên các khu vực.
- Biểu đồ giúp xác định **khu vực có doanh thu cao nhất và thấp nhất** để hỗ trợ định hướng chiến lược kinh doanh.

```python
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Tính tổng doanh thu theo khu vực (đơn vị triệu)
region_sales = (
    df
    .groupby('us_region')['revenue']
    .sum()
    .sort_values(ascending=False)  # Để các thanh được sắp xếp từ trên xuống dưới
    / 1e6                         # Chuyển đổi sang đơn vị triệu
)

# 2. Vẽ biểu đồ
plt.figure(figsize=(10, 4))

sns.barplot(
    x=region_sales.values,
    y=region_sales.index,
    palette='Greens_r'          # Màu xanh lá từ đậm đến nhạt
)

# 3. Định dạng biểu đồ
plt.title('Total Sales by US Region', fontsize=16, pad=12)
plt.xlabel('Total Sales (in Millions USD)', fontsize=12)
plt.ylabel('US Region', fontsize=12)
plt.xticks(rotation=0)

sns.despine(left=True, bottom=True)

plt.tight_layout()
plt.show()
```

### 💡 INSIGHTS

- **West** dẫn đầu với khoảng **360M USD doanh thu** (≈ **35% tổng doanh thu**), khẳng định vị thế là thị trường chủ lực.
- **South và Midwest** mỗi khu vực đóng góp hơn **320M USD** (≈ **32%**), cho thấy nhu cầu mạnh và ổn định tại các khu vực miền Trung.
- **Northeast** đứng cuối với khoảng **210M USD** (≈ **20%**), cho thấy vẫn còn nhiều dư địa để tăng trưởng và cần được đầu tư có trọng điểm.
- **Hành động:** Tập trung thu hẹp khoảng cách tại **Northeast** thông qua các chương trình khuyến mãi phù hợp với thị trường địa phương và xây dựng quan hệ đối tác chiến lược, đồng thời tiếp tục duy trì các chiến lược bán hàng hiệu quả trên toàn quốc.

---

## 🔹 10. TỔNG DOANH THU THEO BANG (BẢN ĐỒ CHOROPLETH)

**🎯 Mục tiêu:** Trực quan hóa **phân bố doanh thu theo địa lý** để xác định các bang có hiệu suất cao và thấp, đồng thời phát hiện **khoảng cách về hiệu quả giữa các khu vực**.

**📊 Biểu đồ:** Bản đồ Choropleth của Hoa Kỳ (US Choropleth Map)

**📈 Loại phân tích EDA:** Phân tích đơn biến theo không gian địa lý (Univariate Geospatial Analysis)

**📋 Cấu trúc biểu đồ:**
- Các bang được **tô màu theo tổng doanh thu**, tính bằng **triệu USD**, sử dụng dải màu xanh để thể hiện mức độ doanh thu.
- **Chú giải (Legend)** ở bên phải thể hiện thang doanh thu (**M USD**).
- Khi di chuột vào từng bang, **Tooltip** hiển thị doanh thu chính xác của bang đó.
- Bản đồ được **giới hạn trong phạm vi Hoa Kỳ** để dễ dàng quan sát và so sánh sự khác biệt giữa các khu vực.

```python
import plotly.express as px

# 1. Tính tổng doanh thu theo từng bang (đơn vị triệu)
state_sales = (
    df
    .groupby('state')['revenue']
    .sum()
    .reset_index()
)

state_sales['revenue_m'] = state_sales['revenue'] / 1e6  # Chuyển đổi sang đơn vị triệu

# 2. Vẽ bản đồ Choropleth bằng Plotly
fig = px.choropleth(
    state_sales,
    locations='state',            # Cột chứa mã bang
    locationmode='USA-states',    # Cho Plotly biết đây là các bang của Hoa Kỳ
    color='revenue_m',
    scope='usa',
    labels={'revenue_m':'Total Sales (M USD)'},
    color_continuous_scale='Blues',
    hover_data={'revenue_m':':.2f'}  # Hiển thị 2 chữ số thập phân
)

# 3. Điều chỉnh bố cục
fig.update_layout(
    title_text='Total Sales by State',
    margin=dict(l=0, r=0, t=40, b=0),
    coloraxis_colorbar=dict(
        title='Sales (M USD)',
        ticksuffix='M'
    )
)

fig.show()
```

## 🔹 12. TỶ SUẤT LỢI NHUẬN TRUNG BÌNH THEO KÊNH

**🎯 Mục tiêu:** So sánh **tỷ suất lợi nhuận trung bình** giữa các kênh bán hàng để xác định **kênh có khả năng sinh lời cao nhất và thấp nhất**.

**📊 Biểu đồ:** Biểu đồ thanh (Bar Chart)

**📈 Loại phân tích EDA:** Phân tích hai biến (Bivariate Analysis)

**📋 Cấu trúc biểu đồ:**
- Sử dụng **các cột dọc** để thể hiện tỷ suất lợi nhuận trung bình của từng kênh.
- Hiển thị **nhãn dữ liệu (%)** trên mỗi cột để dễ dàng so sánh.
- Các kênh được **sắp xếp theo thứ tự tỷ suất lợi nhuận giảm dần**.
- Biểu đồ giúp xác định **kênh bán hàng có hiệu quả sinh lời cao nhất và thấp nhất**.

```python
# 1️⃣ Tính tỷ suất lợi nhuận trung bình (%) cho từng kênh
channel_margin = (
    df.groupby('channel')['profit_margin_pct']  # Nhóm dữ liệu theo kênh bán hàng
      .mean()                                  # Tính tỷ suất lợi nhuận trung bình (%)
      .sort_values(ascending=False)            # Sắp xếp các kênh từ tỷ suất cao nhất đến thấp nhất
)

# 2️⃣ Thiết lập kích thước biểu đồ để dễ quan sát
plt.figure(figsize=(6, 4))

# 3️⃣ Vẽ biểu đồ thanh thể hiện tỷ suất lợi nhuận trung bình theo kênh
ax = sns.barplot(
    x=channel_margin.index,    # Trục X: tên các kênh
    y=channel_margin.values,   # Trục Y: giá trị tỷ suất lợi nhuận trung bình
    palette='coolwarm'         # Bảng màu cho các thanh
)

# 4️⃣ Thêm tiêu đề và nhãn cho các trục
plt.title('Average Profit Margin by Channel')  # Tiêu đề chính
plt.xlabel('Sales Channel')                    # Nhãn trục X
plt.ylabel('Avg Profit Margin (%)')             # Nhãn trục Y

# 5️⃣ Thêm nhãn hiển thị chính xác tỷ suất lợi nhuận trên từng thanh
for i, v in enumerate(channel_margin.values):
    ax.text(
        i,                  # Vị trí X (chỉ số của thanh)
        v + 0.5,            # Vị trí Y (chiều cao của thanh + một khoảng nhỏ)
        f"{v:.2f}%",        # Nhãn hiển thị tỷ lệ phần trăm với 2 chữ số thập phân
        ha='center',        # Căn giữa văn bản theo chiều ngang
        fontweight='bold'   # In đậm văn bản để dễ đọc
    )

# 6️⃣ Điều chỉnh bố cục để tránh nội dung bị cắt và hiển thị biểu đồ
plt.tight_layout()
plt.show()
```

### 💡 INSIGHTS

- **Export** dẫn đầu với tỷ suất lợi nhuận trung bình **37.93%**, theo sát là **Distributor (37.56%)** và **Wholesale (37.09%)**.
- Mức chênh lệch rất nhỏ (**< 0.2%**) cho thấy **khả năng sinh lời ổn định và nhất quán** giữa tất cả các kênh.
- Sự đồng đều này cho thấy doanh nghiệp đang có **khả năng kiểm soát chi phí và duy trì mức giá hiệu quả** trên toàn bộ các kênh bán hàng.
- Để tối đa hóa lợi nhuận, nên **thúc đẩy tăng trưởng sản lượng ở kênh Export**, đồng thời duy trì **hiệu quả hoạt động tại Distributor và Wholesale**.

## 🔹 14. PHÂN KHÚC KHÁCH HÀNG: DOANH THU SO VỚI TỶ SUẤT LỢI NHUẬN

**🎯 Mục tiêu:** Phân khúc khách hàng dựa trên **tổng doanh thu** và **tỷ suất lợi nhuận trung bình**, đồng thời thể hiện **số lượng đơn hàng**.

**📊 Biểu đồ:** Biểu đồ bong bóng (Bubble Chart / Scatter Plot với kích thước điểm thay đổi)

**📈 Loại phân tích EDA:** Phân tích đa biến (Multivariate Analysis)

**📋 Cấu trúc biểu đồ:**
- **Trục X:** Tổng doanh thu của khách hàng.
- **Trục Y:** Tỷ suất lợi nhuận trung bình.
- **Kích thước bong bóng:** Số lượng đơn hàng của khách hàng.
- Các điểm dữ liệu giúp xác định những khách hàng có **doanh thu cao, lợi nhuận cao và số lượng đơn hàng lớn**, từ đó hỗ trợ phân khúc và xây dựng chiến lược chăm sóc khách hàng phù hợp.

```python
# Tổng hợp các chỉ số theo từng khách hàng
cust_summary = df.groupby('customer_name').agg(
    total_revenue=('revenue', 'sum'),
    total_profit=('profit', 'sum'),
    avg_margin=('profit_margin_pct', 'mean'),
    orders=('order_number', 'nunique')
)

# Chuyển doanh thu sang đơn vị triệu
cust_summary['total_revenue_m'] = cust_summary['total_revenue'] / 1e6

plt.figure(figsize=(7, 5))

# Biểu đồ bong bóng với doanh thu tính bằng triệu
sns.scatterplot(
    data=cust_summary,
    x='total_revenue_m',        # <-- sử dụng doanh thu tính bằng triệu
    y='avg_margin',
    size='orders',
    sizes=(20, 200),
    alpha=0.7
)

plt.title('Customer Segmentation: Revenue vs. Profit Margin')
plt.xlabel('Total Revenue (Million USD)')  # <-- nhãn đã được cập nhật
plt.ylabel('Avg Profit Margin (%)')

plt.tight_layout()
plt.show()
```

* Phần lớn khách hàng tập trung trong khoảng **6–10 triệu USD** và có tỷ suất lợi nhuận ổn định (**~34–40%**), cho thấy chính sách định giá tương đối nhất quán trong nhóm này.
* Những khách hàng có doanh thu **<6 triệu USD** có mức biến động tỷ suất lợi nhuận lớn nhất (**~33–43%**), cho thấy cấu trúc chi phí hoặc mức chiết khấu ở các khách hàng nhỏ có xu hướng biến động nhiều hơn.
* Kích thước bong bóng (**số lượng đơn hàng**) tăng theo doanh thu, nhưng mức tỷ suất lợi nhuận dường như không bị ảnh hưởng — củng cố nhận định rằng **doanh thu là yếu tố thúc đẩy hiệu quả kinh doanh quan trọng hơn số lượng đơn hàng**.

## 🔹 15. *Heatmap Tương Quan Giữa Các Biến Số*

**🎯 Mục tiêu:** Xác định mối quan hệ giữa các biến số quan trọng để phát hiện khả năng xảy ra **đa cộng tuyến (Multicollinearity)**.

**📊 Biểu đồ:** Heatmap tương quan (Correlation Heatmap)

**📈 Loại phân tích EDA:** Phân tích đa biến (Multivariate Analysis)

**📋 Cấu trúc:** Heatmap có hiển thị các hệ số tương quan, được chú thích trực tiếp trên biểu đồ, cho các cột dữ liệu số được lựa chọn.


```python
# Liệt kê các cột số được sử dụng để tính toán tương quan
num_cols = ['quantity', 'unit_price', 'revenue', 'cost', 'profit']

# Tính ma trận tương quan cho các biến số này
corr = df[num_cols].corr()

# Thiết lập kích thước biểu đồ để dễ quan sát
plt.figure(figsize=(6,4))

# Vẽ heatmap với hệ số tương quan và bảng màu viridis
sns.heatmap(
    corr,           # Dữ liệu: ma trận tương quan
    annot=True,     # Hiển thị các hệ số tương quan trên heatmap
    fmt=".2f",      # Định dạng số với 2 chữ số thập phân
    cmap='viridis'  # Bảng màu cho heatmap
)

# Thêm tiêu đề để cung cấp thông tin ngữ cảnh
plt.title('Correlation Matrix')

# Điều chỉnh bố cục để tránh nội dung bị cắt
plt.tight_layout()

# Hiển thị heatmap
plt.show()
```

### INSIGHTS

* **Lợi nhuận và doanh thu** có mối tương quan rất mạnh (**0.87**), cho thấy khi giá trị doanh thu tăng thì lợi nhuận cũng có xu hướng tăng theo.
* **Đơn giá (Unit Price)** là một yếu tố quan trọng: tương quan **0.91 với doanh thu**, **0.79 với lợi nhuận** và **0.94 với chi phí**, cho thấy quyết định về giá có tác động đồng thời đến cả doanh thu và chi phí.
* **Chi phí (Cost)** có mối tương quan mạnh với doanh thu (**0.85**) nhưng tương quan ở mức vừa phải với lợi nhuận (**0.58**). Điều này cho thấy doanh thu cao thường đi kèm chi phí cao hơn, nhưng biên lợi nhuận vẫn có thể thay đổi.
* **Số lượng (Quantity)** gần như không có tương quan với đơn giá hoặc chi phí (**≈0.00**), và chỉ có mối liên hệ tương đối yếu với doanh thu (**0.34**) và lợi nhuận (**0.30**). Điều này cho thấy **sản lượng là yếu tố thứ yếu so với giá bán**.
